## Deezer API

### DON'T : 
- Spara id:n, inte ljudlänkar. Förhandslyssningslänkarna (preview) är oftast tidsbegränsade, så hämta en färsk länk när frågan visas. Då slipper du dessutom riskera att spara ljud. Spela klippet direkt från Deezers URL och ladda aldrig ner det.
- Metadata i en egen databas: villkoren nämner inte uttryckligen lagring av metadata, men de förbjuder nedladdning av innehåll. För ett privat projekt är det knappast ett problem. Planerar du att publicera quizet eller tjäna pengar på det, fråga Deezer först (developers.deezer.com/support).
- Rate limit: ungefär 50 anrop per 5 sekunder, så låt dlt hämta med en liten paus.
- Deezers varumärke: använd inte loggan utan att följa deras riktlinjer.
