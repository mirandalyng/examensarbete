```python
def load_tracks(path: str = "../../data/serving/mart.parquet") -> pd.DataFrame:
    """Läs in låtarna från Parquet. Körs EN gång vid uppstart."""
    df = pd.read_parquet(path)

    # decade kommer från martet (dbt). Rensa rader utan årtionde
    # och se till att typen är heltal (1980, 1990 ...).
    df = df.dropna(subset=["decade"])
    df["decade"] = df["decade"].astype(int)

    return df.reset_index(drop=True)


def filter_by_decades(df: pd.DataFrame, decades: list[int]) -> pd.DataFrame:
    """Returnera en kopia med bara låtar från de valda årtiondena.

    decades: lista med startår, t.ex. [1980, 1990]. Även ett enda
    årtionde skickas som lista, t.ex. [1980].
    """
    if not decades:
        raise ValueError("Välj minst ett årtionde.")

    return df[df["decade"].isin(decades)].copy()
    

load_tracks()
```

```python
from pathlib import Path

PARQUET = Path("../../data/serving/mart.parquet")


def load_tracks() -> pd.DataFrame:
    """Läser in hela Parquet-filen."""
    return pd.read_parquet(PARQUET)


def get_decades(df: pd.DataFrame) -> list[int]:
    """Alla årtionden som finns i datan, sorterade (till selectbox/multiselect)."""
    return sorted(df["decade"].dropna().astype(int).unique().tolist())


def filter_by_decades(df: pd.DataFrame, decades: int | list[int]) -> pd.DataFrame:
    """
    Returnerar låtar från ett eller flera årtionden.

    decades kan vara ett enskilt årtionde (1980) eller en lista ([1980, 1990]).
    Tom lista eller None ger hela DataFrame:n (inget filter).
    """
    if decades is None:
        return df

    if isinstance(decades, int):
        decades = [decades]

    if len(decades) == 0:
        return df

    return df[df["decade"].isin(decades)]
```

```python
def filter_by_test2(df: pd.DataFrame, genres: str | list[str]) -> pd.DataFrame:
    if genres is None:
        return df

    if isinstance(genres, str):
        genres = [genres]

    if len(genres) == 0:
        return df

    wanted = {g for g in genres}

    mask = df["genres"].apply(
        lambda row: any(g in wanted for g in row)
        if row is not None and len(row) > 0
        else False
    )
    return df[mask]
```

```python
def filter_by_genre(df: pd.DataFrame, genres: str | list[str]) -> pd.DataFrame:
    # No genre selected (None) -> no filtering, return all tracks
    if genres is None:
        return df

    # If the user passed a single genre as a string (e.g. "jazz"),
    # wrap it in a list so the rest of the function can treat everything as a list
    if isinstance(genres, str):
        genres = [genres]

    # Empty list -> no genre selected, so again return all tracks
    if len(genres) == 0:
        return df

    # Build a set of the wanted genres, normalized:
    # A set is used because "x in set" is much faster than "x in list"
    wanted = {g for g in genres}

    # Build a boolean mask (one True/False per row in the DataFrame).
    # Each cell in df["genres"] is itself a list (or numpy array) of genres,
    # e.g. ["hip hop", "funk"], so we check every row separately with apply().
    mask = df["genres"].apply(
        # For a row's genre list, return True if ANY of its genres
        # (normalized the same way) is in the wanted set.
        # This makes the filter an OR: a track matches if it has at least
        # one of the selected genres.
        lambda row: any(g in wanted for g in row)
        # Guard: only do the check if the cell is not None and not empty
        if row is not None and len(row) > 0
        # Tracks with no genres (None or []) can never match, so mark them False
        else False
    )

    # Keep only the rows where the mask is True
    return df[mask]
```
